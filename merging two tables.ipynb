```python
import numpy as np
import pandas as pd
```

```python
match = pd.read_csv('NPL matches.csv')
```

```python
# Load NPL 2024
df24 = pd.read_csv("NPL-2024.csv")

# Load NPL 2025
df25 = pd.read_csv("NPL-2025.csv")
```

```python
print("NPL 2024 Shape:", df24.shape)
print("NPL 2025 Shape:", df25.shape)

display(df24.head())
display(df25.head())
```

```python
# Rename columns in 2024
df24 = df24.rename(columns={
    "batsman": "batter",
    "player_dismissed": "player_out",
    "fielder": "fielders"
})

# Rename columns in 2025
df25 = df25.rename(columns={
    "match_no": "match_id"
})
```

```python
df24["season"] = 2024
df25["season"] = 2025
```

```python
combined_df = pd.concat(
    [df24, df25],
    ignore_index=True
)

print("Combined Dataset Shape:", combined_df.shape)
```

```python
# Get unique seasons in chronological order
seasons = sorted(combined_df["season"].unique())

# Initialize match ID
new_id = 1

# Create mapping dictionary
match_id_mapping = {}

for season in seasons:

    # Get unique matches for each season
    matches = combined_df[
        combined_df["season"] == season
    ]["match_id"].drop_duplicates()

    for old_id in matches:
        match_id_mapping[(season, old_id)] = new_id
        new_id += 1

# Assign new match IDs
combined_df["match_id"] = combined_df.apply(
    lambda row: match_id_mapping[(row["season"], row["match_id"])],
    axis=1
)
```

```python
# Check match IDs by season
combined_df.groupby("season")["match_id"].agg(
    ["min", "max", "nunique"]
)
```

```python
combined_df.to_csv(
    "NPL-2024-2025-combined.csv",
    index=False
)

print("Dataset successfully saved!")
```
